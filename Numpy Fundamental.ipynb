## Array Function

```python
a1 = np.random.random((3,3))
a1 = np.round(a1*100)
a1
```

```python
# min/max/sum/prod
# 0 ->col and 1 -> row
# np.max(a1)
# np.min(a1)
# np.prod(a1) #all elemnt multiply
np.max(a1,axis=1)
np.max(a1,axis=0)
```

```python
# mmean/median/std/var
np.mean(a1)
np.mean(a1,axis=0)
np.median(a1)
np.std(a1)
np.var(a1)
```

```python
#trigonmetric function
np.sin(a1)
```

```python
# dot product
a2 = np.arange(12).reshape(3,4)
a3 = np.arange(12,24).reshape(4,3)
np.dot(a2,a3)
```

```python
# log and exponats
np.log(a1)

np.exp(a1)
```

```python
#  round /flor/ceil
np.round(np.random.random((2,3))*100)
np.floor(np.random.random((2,3))*100)
np.ceil(np.random.random((2,3))*100)
```
