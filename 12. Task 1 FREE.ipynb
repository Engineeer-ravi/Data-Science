### Q4:- Write a program to find the euclidean distance between two coordinates.Take both the coordinates from the user as input.

````python
import math

x1 = float(input("Enter x1 coordinate: "))
y1 = float(input("Enter y1 coordinate: "))

x2 = float(input("Enter x2 coordinate: "))
y2 = float(input("Enter y2 coordinate: "))

distance = math.sqrt((x2 - x1)**2 + (y2 - y1)**2)

print("Euclidean Distance:", distance)
````

### Q5:- Write a program to find the simple interest when the value of principle,rate of interest and time period is provided by the user.


````python
p = float(input("Enter the principal amount: "))
r = float(input("Enter the rate of interest: "))
t = float(input("Enter the time period (in years): "))

si = (p * r * t) / 100

print("Simple Interest =", si)
````

### Q6:- Write a program that will tell the number of dogs and chicken are there when the user will provide the value of total heads and legs.

For example:
Input:
heads -> 4
legs -> 12
<br>
Output:
dogs -> 2
chicken -> 2





````python
# Write your code here
heads = int(input("Enter total number of heads: "))
legs = int(input("Enter total number of legs: "))

dogs = (legs - 2 * heads) // 2
chickens = heads - dogs

print("Number of dogs:", dogs)
print("Number of chickens:", chickens)
````

### Q7:- Write a program to find the sum of squares of first n natural numbers where n will be provided by the user.

````python
n = int(input("Enter the value of n: "))

sum_sq = n * (n + 1) * (2 * n + 1) // 6

print("Sum of squares of first", n, "natural numbers is:", sum_sq)
````

### Q8:- Given the first 2 terms of an Arithmetic Series.Find the Nth term of the series. Assume all inputs are provided by the user.

````python
# Write your code here
a = int(input("Enter the first term: "))
b = int(input("Enter the second term: "))
n = int(input("Enter the value of n: "))

d = b - a
nth_term = a + (n - 1) * d

print("The", n, "th term of the Arithmetic Series is:", nth_term)
````
